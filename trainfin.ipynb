```python
import pandas as pd
from pathlib import Path

# ------------------------------------------------------------------
# CONFIG
# ------------------------------------------------------------------
BASE_DIR  = Path(r"C:\bt23ece064\wavelet\fin\annotated_feature_csvs_all_wavelets")
WAVELETS  = ['bior3.5', 'coif3', 'db4', 'haar', 'morlet']   # 5 wavelets
# train_idx and test_idx are assumed to be defined exactly as you posted

# ------------------------------------------------------------------
# 1. Prepare containers
# ------------------------------------------------------------------
train_dfs = {w: [] for w in WAVELETS}   # wavelet -> list[pd.DataFrame]
test_dfs  = {w: [] for w in WAVELETS}

# ------------------------------------------------------------------
# 2. Helper to load if the file exists
# ------------------------------------------------------------------
def load_if_exists(pid: int, wavelet: str):
    path = BASE_DIR / f"features_patient_{pid:03d}_wavelet_{wavelet}.csv"
    if path.exists():
        return pd.read_csv(path)
    else:
        print(f"⚠️  Missing file for patient {pid:03d}, wavelet {wavelet}")
        return None

# ------------------------------------------------------------------
# 3. Populate the dictionaries
# ------------------------------------------------------------------
for pid in train_idx:
    for w in WAVELETS:
        df = load_if_exists(pid, w)
        if df is not None:
            train_dfs[w].append(df)

for pid in test_idx:
    for w in WAVELETS:
        df = load_if_exists(pid, w)
        if df is not None:
            test_dfs[w].append(df)

# ------------------------------------------------------------------
# 4.  Usage examples
# ------------------------------------------------------------------
# • train_dfs['db4']   → list of DataFrames (one per patient) in the train split
# • test_dfs['haar']   → list of DataFrames in the test split
#
# To concatenate a wavelet’s list into a single DataFrame:
#   train_db4 = pd.concat(train_dfs['db4'], ignore_index=True)
#   test_db4  = pd.concat(test_dfs['db4'],  ignore_index=True)
```

```python
import numpy as np
import pandas as pd
import optuna

from sklearn.svm import SVC
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    roc_auc_score, accuracy_score, f1_score,
    precision_score, recall_score, confusion_matrix
)
from sklearn.preprocessing import MinMaxScaler
from sklearn.decomposition import PCA
from imblearn.over_sampling import SMOTE


def svm_optuna_from_df_lists(
    train_df_list,
    test_df_list,
    label_col = "label",
    channel_col = "channel",
    n_trials = 30,
    k_folds = 3,
    pca_n_components= 0.95,
    threshold = None,
    random_state = 42,
):
    """
    Train / tune SVM on lists of DataFrames.
    Adds one‑hot encoding for the `channel` column (if present).
    """

    # ------------------------------------------------------------------
    # 1) Concatenate & drop all‑NaN columns / any‑NaN rows
    # ------------------------------------------------------------------
    print("Concatenating train and test dataframes...")
    train_df = pd.concat(train_df_list, ignore_index=True)
    test_df  = pd.concat(test_df_list,  ignore_index=True)

    # Align columns between splits before NaN handling
    common_cols = train_df.columns.intersection(test_df.columns)
    train_df = train_df[common_cols].copy()
    test_df  = test_df[common_cols].copy()

    print("Dropping NaN columns that are all NaN in both train and test...")
    # Drop columns that are all NaN in train **and** test
    nan_all_cols = [
        c for c in train_df.columns
        if train_df[c].isna().all() and test_df[c].isna().all()
    ]
    train_df.drop(columns=nan_all_cols, inplace=True)
    test_df.drop(columns=nan_all_cols, inplace=True)

    # ------------------------------------------------------------------
    # 2) One‑hot encode the `channel` column (if present)
    # ------------------------------------------------------------------
    if channel_col in train_df.columns:
        print(f"One-hot encoding the `{channel_col}` column...")
        train_df = pd.get_dummies(train_df, columns=[channel_col], prefix=channel_col, dtype=int)
        test_df  = pd.get_dummies(test_df,  columns=[channel_col], prefix=channel_col, dtype=int)
        # Re‑align again (test may lack a channel present in train or vice‑versa)
        train_df, test_df = train_df.align(test_df, join="outer", axis=1, fill_value=0)

    # ------------------------------------------------------------------
    # 3) Final NaN drop
    # ------------------------------------------------------------------
    print("Dropping any remaining rows with NaN values...")
    train_df.dropna(axis=0, how="any", inplace=True)
    test_df.dropna(axis=0,  how="any", inplace=True)

    # ------------------------------------------------------------------
    # 4) Split X / y
    # ------------------------------------------------------------------
    print("Splitting features and target labels...")
    X_train_raw = train_df.drop(columns=[label_col]).values
    y_train     = train_df[label_col].values
    X_test_raw  = test_df.drop(columns=[label_col]).values
    y_test      = test_df[label_col].values

    # ------------------------------------------------------------------
    # 5) Scale + PCA
    # ------------------------------------------------------------------
    print("Scaling features using MinMaxScaler...")
    scaler = MinMaxScaler().fit(X_train_raw)
    X_train_scaled = scaler.transform(X_train_raw)
    X_test_scaled  = scaler.transform(X_test_raw)

    print(f"Performing PCA with {pca_n_components} components...")
    pca = PCA(n_components=pca_n_components, random_state=random_state).fit(X_train_scaled)
    X_train_pca = pca.transform(X_train_scaled)
    X_test_pca  = pca.transform(X_test_scaled)

    # ------------------------------------------------------------------
    # 6) Optuna objective
    # ------------------------------------------------------------------
    def objective(trial):
        print(f"Running objective function for trial {trial.number}...")
        C      = trial.suggest_float("C", 1e-3, 1e3, log=True)
        kernel = trial.suggest_categorical("kernel", ["linear", "rbf", "poly", "sigmoid"])
        gamma  = trial.suggest_float("gamma", 1e-4, 10.0, log=True) if kernel != "linear" else "scale"
        degree = trial.suggest_int("degree", 2, 5) if kernel == "poly" else 3

        clf = SVC(
            C=C, kernel=kernel,
            gamma=gamma if kernel != "linear" else "scale",
            degree=degree,
            probability=False,
            random_state=random_state,
        )

        skf = StratifiedKFold(n_splits=k_folds, shuffle=True, random_state=random_state)
        aucs, precisions = [], []

        print("Starting cross-validation...")
        for tr_idx, val_idx in skf.split(X_train_pca, y_train):
            print(f"Training indices: {tr_idx[:10]}, Validation indices: {val_idx[:10]}")
            X_tr, X_val = X_train_pca[tr_idx], X_train_pca[val_idx]
            y_tr, y_val = y_train[tr_idx], y_train[val_idx]

            # Apply SMOTE to balance the training data
            X_tr, y_tr = SMOTE(random_state=random_state).fit_resample(X_tr, y_tr)
            print(f"Resampled X_tr shape: {X_tr.shape}, y_tr shape: {y_tr.shape}")

            clf.fit(X_tr, y_tr)
            scores = clf.decision_function(X_val)
            preds  = clf.predict(X_val)

            auc = roc_auc_score(y_val, scores)
            precision = precision_score(y_val, preds, zero_division=0)

            print(f"AUC: {auc}, Precision: {precision}")

            aucs.append(auc)
            precisions.append(precision)

        return 0.2 * np.mean(aucs) + 0.8 * np.mean(precisions)

    print("Creating Optuna study and optimizing...")
    study = optuna.create_study(direction="maximize")
    study.optimize(objective, n_trials=n_trials, n_jobs=1)

    # ------------------------------------------------------------------
    # 7) Train best model on full, SMOTE‑balanced training data
    # ------------------------------------------------------------------
    print("Training best model on the full training dataset...")
    bp = study.best_params
    best_clf = SVC(
        C=bp["C"], kernel=bp["kernel"],
        gamma=bp["gamma"] if bp["kernel"] != "linear" else "scale",
        degree=bp["degree"] if bp["kernel"] == "poly" else 3,
        probability=True,
        random_state=random_state,
    )

    X_train_res, y_train_res = SMOTE(random_state=random_state).fit_resample(X_train_pca, y_train)
    best_clf.fit(X_train_res, y_train_res)

    # ------------------------------------------------------------------
    # 8) Evaluate on test set (threshold auto‑tune optional)
    # ------------------------------------------------------------------
    print("Evaluating best model on the test set...")
    y_probs = best_clf.predict_proba(X_test_pca)[:, 1]

    def metrics_at(t):
        y_pred = (y_probs >= t).astype(int)
        return {
            "Threshold":  t,
            "Accuracy":   accuracy_score(y_test, y_pred),
            "AUROC":      roc_auc_score(y_test, y_probs),
            "F1":         f1_score(y_test, y_pred),
            "Precision":  precision_score(y_test, y_pred, zero_division=0),
            "Recall":     recall_score(y_test, y_pred),
            "Confusion":  confusion_matrix(y_test, y_pred).tolist(),
        }

    if threshold is None:
        best = max((metrics_at(t) for t in np.linspace(0.3, 0.9, 61)), key=lambda m: m["F1"])
    else:
        best = metrics_at(threshold)

    print(f"Best F1 score: {best['F1']} at threshold {best['Threshold']}")

    return {
        "model":   best_clf,
        "study":   study,
        "scaler":  scaler,
        "pca":     pca,
        "metrics": best,
    }
```

```python
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
import optuna
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score, precision_score, confusion_matrix
from sklearn.preprocessing import MinMaxScaler
from sklearn.decomposition import PCA
from imblearn.over_sampling import SMOTE

def train_neural_network(train_df_list, test_df_list, label_col="label", k_folds=3, pca_components=0.95, threshold=0.5, random_state=42):
    """
    Train a neural network with Optuna hyperparameter optimization.
    Returns trained model with scaler, PCA transformer, and metrics.
    """
    
    # Neural Network Class
    class DynamicNN(nn.Module):
        def __init__(self, input_dim, hidden_units, dropout_prob):
            super(DynamicNN, self).__init__()
            layers = []
            prev_dim = input_dim
            
            for hidden_dim in hidden_units:
                layers.extend([
                    nn.Linear(prev_dim, hidden_dim),
                    nn.ReLU(),
                    nn.Dropout(dropout_prob)
                ])
                prev_dim = hidden_dim
            
            layers.extend([nn.Linear(prev_dim, 1), nn.Sigmoid()])
            self.model = nn.Sequential(*layers)

        def forward(self, x):
            return self.model(x)

    # Data Preparation
    train_df = pd.concat(train_df_list, ignore_index=True)
    test_df = pd.concat(test_df_list, ignore_index=True)
    
    common_cols = train_df.columns.intersection(test_df.columns)
    train_df = train_df[common_cols]
    test_df = test_df[common_cols]
    
    # Clean data
    nan_cols = [c for c in train_df.columns if train_df[c].isna().all() and test_df[c].isna().all()]
    train_df = train_df.drop(columns=nan_cols + (["channel"] if "channel" in train_df.columns else []))
    test_df = test_df.drop(columns=nan_cols + (["channel"] if "channel" in test_df.columns else []))
    
    train_df = train_df.dropna()
    test_df = test_df.dropna()
    
    # Split features and labels
    X_train = train_df.drop(columns=[label_col]).values
    y_train = train_df[label_col].values
    X_test = test_df.drop(columns=[label_col]).values
    y_test = test_df[label_col].values
    
    # Scale and apply PCA
    scaler = MinMaxScaler().fit(X_train)
    X_train_scaled = scaler.transform(X_train)
    X_test_scaled = scaler.transform(X_test)
    
    pca = PCA(n_components=pca_components, random_state=random_state)
    X_train_pca = pca.fit_transform(X_train_scaled)
    X_test_pca = pca.transform(X_test_scaled)
    
    # Hyperparameter optimization
    def objective(trial):
        n_layers = trial.suggest_int("n_layers", 2, 4)
        dropout_prob = trial.suggest_float("dropout_prob", 0.0, 0.5)
        lr = trial.suggest_float("lr", 1e-4, 1e-2, log=True)
        batch_size = trial.suggest_categorical("batch_size", [32, 64])
        epochs = trial.suggest_int("epochs", 10, 100)
        hidden_units = [trial.suggest_int(f"units_l{i}", 8, 64) for i in range(n_layers)]
        
        aucs, precisions = [], []
        skf = StratifiedKFold(n_splits=k_folds, shuffle=True, random_state=random_state)
        
        for tr_idx, val_idx in skf.split(X_train_pca, y_train):
            # Apply SMOTE
            smote = SMOTE(random_state=42)
            X_train_fold, y_train_fold = smote.fit_resample(X_train_pca[tr_idx], y_train[tr_idx])
            X_val_fold, y_val_fold = X_train_pca[val_idx], y_train[val_idx]
            
            # Train model
            model = DynamicNN(X_train_pca.shape[1], hidden_units, dropout_prob)
            criterion = nn.BCELoss()
            optimizer = optim.Adam(model.parameters(), lr=lr)
            
            train_data = TensorDataset(torch.tensor(X_train_fold, dtype=torch.float32), 
                                     torch.tensor(y_train_fold, dtype=torch.float32))
            train_loader = DataLoader(train_data, batch_size=batch_size, shuffle=True)
            
            for epoch in range(epochs):
                model.train()
                for X_batch, y_batch in train_loader:
                    optimizer.zero_grad()
                    y_pred = model(X_batch).squeeze()
                    loss = criterion(y_pred, y_batch)
                    loss.backward()
                    optimizer.step()
            
            # Evaluate
            model.eval()
            with torch.no_grad():
                y_pred = model(torch.tensor(X_val_fold, dtype=torch.float32)).squeeze().cpu().numpy()
            
            auc = roc_auc_score(y_val_fold, y_pred)
            precision = precision_score(y_val_fold, (y_pred >= threshold).astype(int), zero_division=0)
            aucs.append(auc)
            precisions.append(precision)
        
        return 0.5 * np.mean(aucs) + 0.5 * np.mean(precisions)
    
    # Optimize and train final model
    study = optuna.create_study(direction="maximize")
    study.optimize(objective, n_trials=10)
    
    best_params = study.best_trial.params
    hidden_units = [best_params[f"units_l{i}"] for i in range(best_params["n_layers"])]
    
    # Apply SMOTE to full training data for final model
    smote = SMOTE(random_state=42)
    X_train_resampled, y_train_resampled = smote.fit_resample(X_train_pca, y_train)
    
    final_model = DynamicNN(X_train_pca.shape[1], hidden_units, best_params["dropout_prob"])
    criterion = nn.BCELoss()
    optimizer = optim.Adam(final_model.parameters(), lr=best_params["lr"])
    
    train_data = TensorDataset(torch.tensor(X_train_resampled, dtype=torch.float32), 
                             torch.tensor(y_train_resampled, dtype=torch.float32))
    train_loader = DataLoader(train_data, batch_size=best_params["batch_size"], shuffle=True)
    
    for epoch in range(best_params["epochs"]):
        final_model.train()
        for X_batch, y_batch in train_loader:
            optimizer.zero_grad()
            y_pred = final_model(X_batch).squeeze()
            loss = criterion(y_pred, y_batch)
            loss.backward()
            optimizer.step()
    
    # Test evaluation
    final_model.eval()
    with torch.no_grad():
        y_test_pred = final_model(torch.tensor(X_test_pca, dtype=torch.float32)).squeeze().cpu().numpy()
    
    final_auc = roc_auc_score(y_test, y_test_pred)
    final_precision = precision_score(y_test, (y_test_pred >= threshold).astype(int), zero_division=0)
    final_confusion_matrix = confusion_matrix(y_test, (y_test_pred >= threshold).astype(int))
    
    return {
        "model": final_model,
        "scaler": scaler,
        "pca": pca,
        "final_metrics": {
            "AUC": final_auc, 
            "Precision": final_precision,
            "Confusion_Matrix": final_confusion_matrix
        }
    }
```
